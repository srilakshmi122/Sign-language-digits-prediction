```python
row = []    # add converted images
label = []   # add output names

for folder in os.listdir():
    new_path = origin+'\\' + folder  # entering into each folder
    os.chdir(new_path)

    for image in os.listdir():
        # print(image)
        ndarray = cv2.imread(image , 0)
        resized = cv2.resize(ndarray, (28, 28))
        flattened = resized.flatten()
        row.append(flattened)
        label.append(folder)
```

```python
df1 = pd.DataFrame(row, dtype = np.uint8)
```

```python
df1.head()
```

```python
df1
```

```python
df2 = pd.DataFrame(label,columns = ["Label"])
df2
```

```python
df = pd.concat([df1, df2], axis = 1)
df
```

```python
X = df.drop(columns = ['Label'])
```

```python
y = df['Label']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train,y_test = train_test_split(X, y,test_size = 0.2, random_state = 23)
```

```python
from sklearn.metrics import accuracy_score
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier()
knn
```

```python
knn.fit(X_train, y_train)
knn
```

```python
y_pred = knn.predict(X_test)
y_pred
```

```python
result = pd.DataFrame({
    "actual":y_test,
    "predicted":y_pred
})
print(result[:21])
```

```python
accuracy_score(y_test, y_pred)
```

```python
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(random_state = 23)
dt.fit(X_train, y_train)
y_pred = dt.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
def prediction(Path):
    ndarray = cv2.imread(Path, 0)
    resized = cv2.resize(ndarray, (28,28))
    flattened_image = resized.flatten()
    return knn.predict([flattened_image])[0]
```
